# 04 Vessel Anomaly

Goals:
- Review fact_vessel_anomalies
- Evidence trails
- Human-review workflow (no illegal labels)


```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine("postgresql://postgres:password@localhost:5433/oceanwatch_db")
anom = pd.read_sql("""
    SELECT vessel_name, vessel_type, risk_score, confidence_score, status, evidence
    FROM fact_vessel_anomalies
    ORDER BY risk_score DESC
""", engine)
anom.head(15)
```

```python
# Human review queue: only flag for review, never auto-label as illegal
review = anom[anom["status"] == "REQUIRES_HUMAN_REVIEW"].copy()
print(f"Cases requiring human review: {len(review)}")
review
```
